Accept 5 paper marks, roll no, name and calculation total & percentage if percentage >=85, student can take addmission else cant take addmission

uing class & object & constructor

```python
class Student:
    def __init__(self, rollno, name, m1, m2, m3, m4, m5):
        self.rollno = rollno
        self.name = name
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
        self.m4 = m4
        self.m5 = m5

    def calculate(self):
        total = self.m1 + self.m2 + self.m3 + self.m4 + self.m5
        percentage = total / 5

        print("\nRoll No:", self.rollno)
        print("Name:", self.name)
        print("Total Marks:", total)
        print("Percentage:", percentage)

        if percentage >= 85:
            print("Student can take admission")
        else:
            print("Student cannot take admission")


rollno = int(input("Enter Roll No: "))
name = input("Enter Name: ")

m1 = int(input("Enter Paper 1 Marks: "))
m2 = int(input("Enter Paper 2 Marks: "))
m3 = int(input("Enter Paper 3 Marks: "))
m4 = int(input("Enter Paper 4 Marks: "))
m5 = int(input("Enter Paper 5 Marks: "))

s = Student(rollno, name, m1, m2, m3, m4, m5)

s.calculate()
```

Single level inheritance

```python
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

obj=B()
obj.showA()
obj.showB()
```

Multi level inheritance

```python
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

class C(B):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC()
```

Hybrid Inheritance

```python
class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

class D(B, C):
    def showD(self):
        print("showD function")

obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```

Hierarchal Inheritance

```python
class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

obj1 = B()
obj2 = C()

obj1.showA()
obj1.showB()

obj2.showA()
obj2.showC()
```
